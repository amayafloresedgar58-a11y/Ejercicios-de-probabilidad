```python

import random


# ============================================================
# EJERCICIO 1
# Simular 10,000 lanzamientos de dos dados
# y estimar la probabilidad de obtener suma 5
# ============================================================

def ejercicio_1():
    lanzamientos = 10000
    favorables = 0

    for _ in range(lanzamientos):
        dado1 = random.randint(1, 6)
        dado2 = random.randint(1, 6)

        if dado1 + dado2 == 5:
            favorables += 1

    probabilidad = favorables / lanzamientos
    clasica = 4 / 36

    print("\n--- EJERCICIO 1 ---")
    print("Lanzamientos:", lanzamientos)
    print("Casos favorables:", favorables)
    print(f"Probabilidad frecuentista: {probabilidad:.4f}")
    print(f"Porcentaje: {probabilidad * 100:.2f}%")
    print(f"Probabilidad clásica: {clasica:.4f}")


# ============================================================
# EJERCICIO 2
# Simular 10,000 PIN de 4 dígitos sin repeticiones
# ============================================================

def generar_pin():
    return [
        random.randint(0, 9)
        for _ in range(4)
    ]


def ejercicio_2():
    simulaciones = 10000
    favorables = 0

    for _ in range(simulaciones):
        pin = generar_pin()

        if len(set(pin)) == 4:
            favorables += 1

    probabilidad = favorables / simulaciones
    clasica = (10 * 9 * 8 * 7) / (10 ** 4)

    print("\n--- EJERCICIO 2 ---")
    print("PIN simulados:", simulaciones)
    print("PIN sin dígitos repetidos:", favorables)
    print(f"Probabilidad frecuentista: {probabilidad:.4f}")
    print(f"Porcentaje: {probabilidad * 100:.2f}%")
    print(f"Probabilidad clásica: {clasica:.4f}")


# ============================================================
# EJERCICIO 3
# Simular 20,000 lanzamientos de tres dados
# y comparar las sumas 9 y 10
# ============================================================

def ejercicio_3():
    lanzamientos = 20000
    suma_9 = 0
    suma_10 = 0

    for _ in range(lanzamientos):
        dado1 = random.randint(1, 6)
        dado2 = random.randint(1, 6)
        dado3 = random.randint(1, 6)

        suma = dado1 + dado2 + dado3

        if suma == 9:
            suma_9 += 1
        elif suma == 10:
            suma_10 += 1

    p9 = suma_9 / lanzamientos
    p10 = suma_10 / lanzamientos

    print("\n--- EJERCICIO 3 ---")
    print("Lanzamientos:", lanzamientos)
    print("Resultados con suma 9:", suma_9)
    print("Resultados con suma 10:", suma_10)
    print(f"Probabilidad de suma 9: {p9:.4f}")
    print(f"Probabilidad de suma 10: {p10:.4f}")

    if suma_9 > suma_10:
        print("En esta simulación salió más la suma 9.")
    elif suma_10 > suma_9:
        print("En esta simulación salió más la suma 10.")
    else:
        print("Ambas sumas aparecieron la misma cantidad de veces.")


# ============================================================
# EJERCICIO 4
# Urna con 3 bolas rojas y 2 azules
# Realizar 2,000 extracciones con reemplazo
# ============================================================

def ejercicio_4():
    urna = ["roja", "roja", "roja", "azul", "azul"]
    extracciones = 2000
    azules = 0

    for _ in range(extracciones):
        bola = random.choice(urna)

        if bola == "azul":
            azules += 1

    probabilidad = azules / extracciones
    teorica = 2 / 5

    print("\n--- EJERCICIO 4 ---")
    print("Extracciones:", extracciones)
    print("Bolas azules obtenidas:", azules)
    print(f"Probabilidad frecuentista: {probabilidad:.4f}")
    print(f"Porcentaje: {probabilidad * 100:.2f}%")
    print(f"Probabilidad teórica: {teorica:.4f}")


# ============================================================
# EJERCICIO 5
# Simular 10,000 series de 20 volados
# Calcular la probabilidad de obtener al menos
# 5 águilas consecutivas en una serie
# ============================================================

def lanzar_moneda():
    return random.choice(["águila", "sol"])


def hay_cinco_aguilas_consecutivas(serie):
    racha = 0

    for resultado in serie:
        if resultado == "águila":
            racha += 1

            if racha >= 5:
                return True
        else:
            racha = 0

    return False


def ejercicio_5():
    series = 10000
    longitud = 20
    favorables = 0

    for _ in range(series):
        serie = [
            lanzar_moneda()
            for _ in range(longitud)
        ]

        if hay_cinco_aguilas_consecutivas(serie):
            favorables += 1

    probabilidad = favorables / series

    print("\n--- EJERCICIO 5 ---")
    print("Series simuladas:", series)
    print("Series con 5 águilas consecutivas:", favorables)
    print(f"Probabilidad estimada: {probabilidad:.4f}")
    print(f"Porcentaje: {probabilidad * 100:.2f}%")


# ============================================================
# EJERCICIO 6 — BONUS
# Simular la llegada de 500 clientes
# Llegadas: cada 1 a 5 minutos
# Atención: de 2 a 4 minutos
# Calcular la probabilidad de esperar más de 10 minutos
# ============================================================

def ejercicio_6():
    clientes = 500
    tiempo_llegada = 0
    fin_atencion_anterior = 0
    esperas_mayores_10 = 0

    for _ in range(clientes):
        # Tiempo entre la llegada de dos clientes
        intervalo = random.randint(1, 5)
        tiempo_llegada += intervalo

        # Tiempo de atención del cajero
        duracion_atencion = random.randint(2, 4)

        # El cliente comienza cuando llega o cuando
        # termina de atenderse el cliente anterior
        inicio_atencion = max(
            tiempo_llegada,
            fin_atencion_anterior
        )

        espera = inicio_atencion - tiempo_llegada

        if espera > 10:
            esperas_mayores_10 += 1

        fin_atencion_anterior = (
            inicio_atencion + duracion_atencion
        )

    probabilidad = esperas_mayores_10 / clientes

    print("\n--- EJERCICIO 6: BONUS ---")
    print("Clientes simulados:", clientes)
    print("Clientes que esperaron más de 10 minutos:",
          esperas_mayores_10)
    print(f"Probabilidad estimada: {probabilidad:.4f}")
    print(f"Porcentaje: {probabilidad * 100:.2f}%")


# ============================================================
# MENÚ PRINCIPAL
# ============================================================

def menu():
    ejercicios = {
        "1": ejercicio_1,
        "2": ejercicio_2,
        "3": ejercicio_3,
        "4": ejercicio_4,
        "5": ejercicio_5,
        "6": ejercicio_6
    }

    while True:
        print("\n======================================")
        print("   PROBABILIDAD FRECUENTISTA")
        print("======================================")
        print("1. Dos dados: suma 5")
        print("2. PIN de 4 dígitos sin repeticiones")
        print("3. Tres dados: suma 9 y 10")
        print("4. Urna con bolas rojas y azules")
        print("5. Veinte volados: cinco águilas seguidas")
        print("6. Bonus: fila de clientes")
        print("7. Ejecutar todos los ejercicios")
        print("0. Salir")

        opcion = input("Selecciona una opción: ")

        if opcion == "0":
            print("Programa finalizado.")
            break

        elif opcion == "7":
            for ejercicio in ejercicios.values():
                ejercicio()

        elif opcion in ejercicios:
            ejercicios[opcion]()

        else:
            print("Opción no válida. Intenta de nuevo.")


if __name__ == "__main__":
    menu()
```
