#**EXPERIMENTOS**

##Experimento determinista


**Un experimento determinista es aquel que produce siempre el mismo resultado cuando se repite bajo las mismas condiciones iniciales. Es decir, si las variables que intervienen en el experimento se mantienen constantes, el resultado será siempre el mismo.**

***Ejemplo de Area***
◀


```python
import sympy

def area(radio):
    return sympy.pi * (radio ** 2)
```

```python
area(10)
```

```python
def area_numerco(radio):
  return 3.1416 * (radio ** 2)
```

```python
area_numerco(5)
```

##Experimenro aleatorio

**En contraste, un experimento aleatorio es aquel que, aún cuando se repita bajo las mismas condiciones, no produce siempre el mismo resultado y no es posible predecir con certeza cuál ocurrirá.**

***Ejemplo de mondea***
◀

```python
import random

moneda_lados = ["aguila", "sol"]

def lanzar_moneda():
    return random.choice(moneda_lados)

lanzar_moneda()
```

```python
import random

dado_caras = ["cara 1", "cara 2", "cara 3", "cara 4", "cara 5", "cara 6"]

def lanzar_dado():
    return random.choice(dado_caras)

lanzar_dado()
```
