```python
1
import random
import math
from itertools import product, permutations


# ============================================================
# EJERCICIO 1
# Probabilidad de que dos dados sumen exactamente 5
# ============================================================

def ejercicio_1():
    resultados = list(product(range(1, 7), repeat=2))
    favorables = [par for par in resultados if sum(par) == 5]

    probabilidad = len(favorables) / len(resultados)

    print("\n--- EJERCICIO 1 ---")
    print("Resultados favorables:", favorables)
    print("Casos favorables:", len(favorables))
    print("Casos posibles:", len(resultados))
    print(f"Probabilidad: {probabilidad:.4f}")
    print(f"Porcentaje: {probabilidad * 100:.2f}%")


# ============================================================
# EJERCICIO 2
# Probabilidad de que una contraseña de 3 dígitos
# no tenga dígitos repetidos
# ============================================================

def ejercicio_2():
    total = 10 ** 3
    favorables = 10 * 9 * 8

    probabilidad = favorables / total

    print("\n--- EJERCICIO 2 ---")
    print("Casos posibles:", total)
    print("Casos favorables:", favorables)
    print(f"Probabilidad: {probabilidad:.4f}")
    print(f"Porcentaje: {probabilidad * 100:.2f}%")


# ============================================================
# EJERCICIO 3
# Probabilidad de que la canción favorita quede primera
# y la que menos gusta quede última
# ============================================================

def ejercicio_3():
    canciones = list(range(8))
    ordenes_totales = math.factorial(8)
    ordenes_favorables = math.factorial(6)

    probabilidad = ordenes_favorables / ordenes_totales

    print("\n--- EJERCICIO 3 ---")
    print("Canciones distintas:", len(canciones))
    print("Órdenes posibles:", ordenes_totales)
    print("Órdenes favorables:", ordenes_favorables)
    print(f"Probabilidad: {probabilidad:.6f}")
    print(f"Porcentaje: {probabilidad * 100:.4f}%")


# ============================================================
# EJERCICIO 4
# Comparar la probabilidad de obtener suma 9 o suma 10
# al lanzar tres dados
# ============================================================

def ejercicio_4():
    resultados = list(product(range(1, 7), repeat=3))

    suma_9 = [terna for terna in resultados if sum(terna) == 9]
    suma_10 = [terna for terna in resultados if sum(terna) == 10]

    casos_totales = len(resultados)
    prob_9 = len(suma_9) / casos_totales
    prob_10 = len(suma_10) / casos_totales

    print("\n--- EJERCICIO 4 ---")
    print("Casos posibles:", casos_totales)
    print("Resultados con suma 9:", len(suma_9))
    print("Resultados con suma 10:", len(suma_10))
    print(f"Probabilidad de suma 9: {prob_9:.4f}")
    print(f"Probabilidad de suma 10: {prob_10:.4f}")

    diferencia = abs(len(suma_9) - len(suma_10))

    if len(suma_9) > len(suma_10):
        print(f"La suma 9 es más probable por {diferencia} resultados.")
    elif len(suma_10) > len(suma_9):
        print(f"La suma 10 es más probable por {diferencia} resultados.")
    else:
        print("Ambas sumas tienen la misma probabilidad.")


# ============================================================
# EJERCICIO 5
# Probabilidad de que nadie saque su propio regalo
# en un amigo secreto de cuatro personas
# ============================================================

def ejercicio_5():
    personas = [0, 1, 2, 3]
    repartos = list(permutations(personas))

    favorables = [
        reparto
        for reparto in repartos
        if all(personas[i] != reparto[i] for i in range(4))
    ]

    total = len(repartos)
    casos_favorables = len(favorables)
    probabilidad = casos_favorables / total

    nombres = ["Ana", "Bruno", "Carla", "Diego"]

    print("\n--- EJERCICIO 5 ---")
    print("Personas:", nombres)
    print("Repartos posibles:", total)
    print("Repartos donde nadie recibe su propio regalo:",
          casos_favorables)

    print("Ejemplos de repartos favorables:")
    for reparto in favorables:
        asignacion = [
            f"{nombres[i]} recibe el regalo de {nombres[reparto[i]]}"
            for i in range(4)
        ]
        print(" - " + "; ".join(asignacion))

    print(f"Probabilidad: {probabilidad:.4f}")
    print(f"Porcentaje: {probabilidad * 100:.2f}%")


# ============================================================
# PROGRAMA PRINCIPAL
# ============================================================

def menu():
    while True:
        print("\n===================================")
        print("   EJERCICIOS DE PROBABILIDAD")
        print("===================================")
        print("1. Lanzamiento de dos dados")
        print("2. Contraseña sin dígitos repetidos")
        print("3. Orden de ocho canciones")
        print("4. Comparación de sumas con tres dados")
        print("5. Problema del amigo secreto")
        print("6. Ejecutar todos los ejercicios")
        print("0. Salir")

        opcion = input("\nSelecciona una opción: ")

        if opcion == "1":
            ejercicio_1()
        elif opcion == "2":
            ejercicio_2()
        elif opcion == "3":
            ejercicio_3()
        elif opcion == "4":
            ejercicio_4()
        elif opcion == "5":
            ejercicio_5()
        elif opcion == "6":
            for ejercicio in [
                ejercicio_1,
                ejercicio_2,
                ejercicio_3,
                ejercicio_4,
                ejercicio_5
            ]:
                ejercicio()
        elif opcion == "0":
            print("Programa finalizado.")
            break
        else:
            print("Opción no válida. Intenta de nuevo.")


if __name__ == "__main__":
    menu()
```
