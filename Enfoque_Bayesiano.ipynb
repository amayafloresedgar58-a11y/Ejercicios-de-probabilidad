```python

from itertools import product


# ============================================================
# 1. PROBABILIDAD CONDICIONAL
# ============================================================

def condicional_1():
    # Obtener un 2 dado que salió un número par
    dados_pares = [2, 4, 6]
    favorables = [n for n in dados_pares if n == 2]

    p = len(favorables) / len(dados_pares)

    print("\nEjercicio 1: Dado y número par")
    print(f"Probabilidad: {p:.4f} ({p * 100:.2f}%)")


def condicional_2():
    # Tres dados: suma 9, dado que las caras son distintas
    resultados = list(product(range(1, 7), repeat=3))

    caras_distintas = [
        r for r in resultados
        if len(set(r)) == 3
    ]

    favorables = [
        r for r in caras_distintas
        if sum(r) == 9
    ]

    p = len(favorables) / len(caras_distintas)

    print("\nEjercicio 2: Tres dados y suma 9")
    print("Casos con caras distintas:", len(caras_distintas))
    print("Casos favorables:", len(favorables))
    print(f"Probabilidad: {p:.4f} ({p * 100:.2f}%)")


def condicional_3():
    # De 100 días, 40 son nublados y en 30 de esos llueve
    dias_nublados = 40
    dias_nublados_con_lluvia = 30

    p = dias_nublados_con_lluvia / dias_nublados

    print("\nEjercicio 3: Lluvia dado que está nublado")
    print(f"Probabilidad: {p:.4f} ({p * 100:.2f}%)")


# ============================================================
# 2. TEOREMA DE PROBABILIDAD TOTAL
# ============================================================

def total_1():
    # Clientes que ven un anuncio y realizan una compra
    p_anuncio = 0.70
    p_compra_anuncio = 0.30
    p_sin_anuncio = 0.30
    p_compra_sin_anuncio = 0.10

    p = (
        p_anuncio * p_compra_anuncio
        + p_sin_anuncio * p_compra_sin_anuncio
    )

    print("\nEjercicio 4: Anuncio y compra")
    print(f"Probabilidad de compra: {p:.4f} ({p * 100:.2f}%)")


def total_2():
    # Probabilidad de lluvia en días nublados y no nublados
    p_nublado = 0.40
    p_lluvia_nublado = 0.75
    p_no_nublado = 0.60
    p_lluvia_no_nublado = 0.10

    p = (
        p_nublado * p_lluvia_nublado
        + p_no_nublado * p_lluvia_no_nublado
    )

    print("\nEjercicio 5: Probabilidad total de lluvia")
    print(f"Probabilidad: {p:.4f} ({p * 100:.2f}%)")


def total_3():
    # Dos cajas de audífonos elegidas con igual probabilidad
    # Caja 1: 2 funcionan y 6 fallan
    # Caja 2: 5 funcionan y 3 fallan
    p_caja_1 = 0.50
    p_funciona_1 = 2 / 8

    p_caja_2 = 0.50
    p_funciona_2 = 5 / 8

    p = (
        p_caja_1 * p_funciona_1
        + p_caja_2 * p_funciona_2
    )

    print("\nEjercicio 6: Audífonos que funcionan")
    print(f"Probabilidad: {p:.4f} ({p * 100:.2f}%)")


# ============================================================
# 3. TEOREMA DE BAYES
# ============================================================

def bayes_1():
    # Probabilidad de que un correo sea spam dado que contiene
    # la palabra "oferta"
    p_spam = 0.20
    p_oferta_spam = 0.80
    p_oferta_deseado = 0.10
    p_deseado = 1 - p_spam

    p_oferta = (
        p_spam * p_oferta_spam
        + p_deseado * p_oferta_deseado
    )

    p_spam_oferta = (
        p_spam * p_oferta_spam / p_oferta
    )

    print("\nEjercicio 7: Correo spam")
    print("Probabilidad de spam dado que contiene 'oferta':")
    print(f"{p_spam_oferta:.4f} ({p_spam_oferta * 100:.2f}%)")


def bayes_2():
    # Probabilidad del navegador usado dado que hubo una venta
    navegadores = {
        "Chrome": {"uso": 0.40, "conversion": 0.03},
        "Safari": {"uso": 0.35, "conversion": 0.04},
        "Otros": {"uso": 0.25, "conversion": 0.02}
    }

    # Probabilidad total de que se registre una venta
    p_venta = sum(
        datos["uso"] * datos["conversion"]
        for datos in navegadores.values()
    )

    print("\nEjercicio 8: Navegador y venta")
    print(f"Probabilidad total de venta: {p_venta:.4f}")

    posteriores = {}

    for nombre, datos in navegadores.items():
        p = datos["uso"] * datos["conversion"] / p_venta
        posteriores[nombre] = p

        print(
            f"{nombre}: {p:.4f} "
            f"({p * 100:.2f}%)"
        )

    mas_probable = max(posteriores, key=posteriores.get)
    print("Navegador más probable:", mas_probable)


# ============================================================
# MENÚ PRINCIPAL
# ============================================================

def menu():
    ejercicios = {
        "1": condicional_1,
        "2": condicional_2,
        "3": condicional_3,
        "4": total_1,
        "5": total_2,
        "6": total_3,
        "7": bayes_1,
        "8": bayes_2
    }

    while True:
        print("\n======================================")
        print("    EJERCICIOS DE PROBABILIDAD")
        print("======================================")
        print("1. Probabilidad condicional: dado")
        print("2. Probabilidad condicional: tres dados")
        print("3. Probabilidad condicional: lluvia")
        print("4. Probabilidad total: anuncio")
        print("5. Probabilidad total: lluvia")
        print("6. Probabilidad total: audífonos")
        print("7. Teorema de Bayes: correo spam")
        print("8. Teorema de Bayes: navegador")
        print("9. Ejecutar todos los ejercicios")
        print("0. Salir")

        opcion = input("Selecciona una opción: ")

        if opcion == "0":
            print("Programa finalizado.")
            break
        elif opcion == "9":
            for ejercicio in ejercicios.values():
                ejercicio()
        elif opcion in ejercicios:
            ejercicios[opcion]()
        else:
            print("Opción no válida.")


if __name__ == "__main__":
    menu()
```
